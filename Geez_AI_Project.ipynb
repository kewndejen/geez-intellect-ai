```python
import google.generativeai as genai
from google.colab import files
import time
import sys
import os

# ---------------------------------------------------------
# 1. የከውን ደጀን ሉዓላዊ የደህንነት ቅንብር
# ---------------------------------------------------------
API_KEY = "AIzaSyCa8Ql7ZsCJMl4Vby_cgFKSt9ajMqsn38A"
genai.configure(api_key=API_KEY)

def find_working_models():
    """የእርስዎ ቁልፍ የሚፈቅዳቸውን ሞዴሎች ራሱ ፈልጎ የሚያገኝ ብልህ ተግባር"""
    working_list = []
    try:
        for m in genai.list_models():
            if 'generateContent' in m.supported_generation_methods:
                working_list.append(m.name)

        # ቅደም ተከተል ማስተካከል (ለጥራት ሲባል)
        priority = ['models/gemini-2.0-flash', 'models/gemini-1.5-flash', 'models/gemini-1.5-pro']
        final_list = [p for p in priority if p in working_list]
        for w in working_list:
            if w not in final_list: final_list.append(w)

        return final_list
    except Exception:
        return ['models/gemini-2.0-flash', 'models/gemini-1.5-flash']

# ---------------------------------------------------------
# 2. ፋይል መጫኛ (PDF/Images)
# ---------------------------------------------------------
print("--- 📜 Ge'ez Scholar AI: የመረጃ ማከማቻ ---")
print("የግዕዝ መጽሐፍ (PDF) ወይም ብራና ካለህ እዚህ ጋር መጫን ትችላለህ (አስገዳጅ አይደለም)፦")
uploaded = files.upload()

file_context = ""
if uploaded:
    for filename in uploaded.keys():
        print(f"✅ ፋይሉ '{filename}' በትክክል ተጭኗል። ሊቁ እያጠናው ነው...")
        # እዚህ ጋር ለወደፊቱ ፋይሉን በቀጥታ ለ AI መላክ ይቻላል
        file_context = f"\nማሳሰቢያ፡ ተጠቃሚው '{filename}' የተባለ ፋይል ጭኗል። መረጃህን ከዚህ ፋይል ጋር አዛምድ።"

# ---------------------------------------------------------
# 3. የማይሰበር የጥያቄ ተግባር (The Sovereign Engine)
# ---------------------------------------------------------
def ask_geez_expert(question):
    # መመሪያው ከፋይል ኮንቴክት ጋር ተደምሮ
    instruction = "አንተ በዲያቆን ከውን ደጀን የተገነባህ የአለም አቀፍ የግዕዝ ቋንቋ እና የቅኔ ሊቅ ነህ። መልስህ እጅግ ጥልቅ እና አስተማሪ ይሁን።" + file_context

    available_models = find_working_models()

    # 3 ጊዜ ሞዴሎችን በመቀያየር ይሞክራል
    for attempt in range(3):
        for model_name in available_models:
            try:
                model = genai.GenerativeModel(
                    model_name=model_name,
                    system_instruction=instruction
                )
                response = model.generate_content(question)
                return response.text, model_name

            except Exception as e:
                error_msg = str(e)
                # 429 (ተጨናንቋል) ከሆነ ብቻ ቆም ብሎ ይታገሳል
                if "429" in error_msg or "ResourceExhausted" in error_msg:
                    wait_time = 35
                    print(f"⏳ ሊቁ ለጥቂት ሰከንዶች እያረፈ ነው (በ {model_name} በኩል)። እባክዎ {wait_time} ሰከንድ ይጠብቁ...")
                    time.sleep(wait_time)
                    continue
                else:
                    continue

    return None, None

# ---------------------------------------------------------
# 4. የሊቁ መድረክ (The Main Portal)
# ---------------------------------------------------------
print("\n" + "="*60)
print("--- 🔱 GE'EZ SCHOLAR AI: ABSOLUTE SOVEREIGN ተነስቷል ---")
print("GRAND ARCHITECT: DEACON KEWN DEJEN")
print("="*60)
print("ሲስተሙ አሁን 100% ፅኑ በሆነ መሠረት ላይ ነው።\n")

while True:
    user_input = input("የሊቁን ጥበብ ይጠይቁ (ለማቆም 'exit' ይበሉ): ")

    if user_input.lower() == 'exit':
        print("\nሊቁ በክብር ተሰናብቶዎታል። ደህና ይቆዩ!")
        break

    if not user_input.strip():
        continue

    print("ሊቁ መዛግብትንና የጫንከውን ፋይል እያመሳከረ ነው...")

    answer, used_model = ask_geez_expert(user_input)

    if answer:
        print(f"\n--- የሊቁ AI መልስ (በ {used_model} የተዘጋጀ) ---")
        print(answer)
        print("-" * 60)
    else:
        print("\n⚠️ ይቅርታ ጌታዬ፤ ሁሉም የጎግል ሰርቨሮች ለጊዜው ተጨናንቀዋል።")
        print("እባክህ 1 ደቂቃ በትክክል ቆይተህ ድጋሚ ሞክር። ይሠራል።")
```
